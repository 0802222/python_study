```python
# numpy를 이용한 거듭 제곱 연산
j = np.arange(5)
print(j)

print(2 ** (j + 1) - j)

"""
풀이
1. j + 1을 계산하면 [1, 2, 3, 4, 5]가 된다.
2. 2 ** (j + 1)을 계산하면 [2, 4, 8, 16, 32]가 된다. (2가 밑, j + 1이 지수)
3. 결과에 j를 빼면 [2, 3, 6, 13, 28]이 된다.
       [2  4  8 16 32]
     - [0  1  2  3  4]
    -------------------
       [2  3  6 13 28]
"""
```

```python
# 순수 파이썬을 이용한 거듭 제곱 연산
j = list(range(5))
print(j)

print([2 ** (i + 1) - i for i in j])
```

```python
# 두 방법을 측정해서 비교해보자 (numpy vs pure python)

import time
import numpy as np

# 데이터 준비: 측정에서 제외
j_python = list(range(5))
j_numpy = np.arange(5)

# 순수 파이썬 버전
start_time = time.perf_counter()

result_python = [2 ** (i + 1) - i for i in j_python]

end_time = time.perf_counter()
python_time = end_time - start_time

# NumPy 버전
start_time = time.perf_counter()

result_numpy = 2 ** (j_numpy + 1) - j_numpy

end_time = time.perf_counter()
numpy_time = end_time - start_time

# 출력은 측정이 끝난 뒤에!
print("파이썬 결과:", result_python)
print("NumPy 결과:", result_numpy)

print(f"파이썬 실행 시간: {python_time:.9f}초")
print(f"NumPy 실행 시간: {numpy_time:.9f}초")

# numpy가 훨씬 빠른 것을 확인할 수 있다. (특히 데이터가 많을수록 차이가 커진다.)
```

```python
# array 곱셈은 행렬 곱셈(a @ b)이 아니라 elementwise 곱셈이다.
# elementwise 곱셈은 각 요소끼리 곱하는 것을 의미한다. (a * b)

a = np.arange(10000)
%timeit a + 1
```

```python
l = range(10000)
%timeit [i + 1 for i in l]
```

```python
# Consider these examples
# 1x1 = 1 (각 위치에서 곱셈)
c = np.ones((3, 3))
print(c)

c * c
```

```python
# Matrix multiplication
# 1×1 + 1×1 + 1×1 = 3 (각 위치에서 곱셈 후 합산)
c @ c
```

Other operations

```python
# Comparisons (각 요소끼리 비교)
a = np.array([1, 2, 3, 4])
b = np.array([4, 2, 2, 4])
a == b
```

```python
a > b
```

```python
# Array-wise comparisons
a = np.array([1, 2, 3, 4])
b = np.array([4, 2, 2, 4])
c = np.array([1, 2, 3, 4])
print("a == b:", np.array_equal(a, b))
print("a == c:", np.array_equal(a, c))
```

```python
# Logical operations (각 요소끼리 논리 연산)
a = np.array([1, 1, 0, 0], dtype=bool)
b = np.array([1, 0, 1, 0], dtype=bool)
print("a or b:", np.logical_or(a, b))
print("a and b:", np.logical_and(a, b))
```

```python
# Transcendental functions (각 요소에 대해 삼각함수, 지수함수, 로그함수 등 적용)
# sin() : radians 단위로 입력된 각도에 대한 사인값을 반환한다. 
a = np.arange(5)
print("a:", a)
print("sin(a):", np.sin(a))
```

```python
# exp() : 자연로그의 밑 e(e ≈ 2.71828)의 거듭 제곱을 계산한다. (밑:e, 지수:각 요소)
print("exp(a):", np.exp(a))
```

```python
# log() : 자연로그를 계산한다. (e를 몇제곱해야 하는지 계산, 밑: e, 지수:각 요소)
print("log(exp(a)):", np.log(np.exp(a)))
```

```python
# Shape mismatches
a = np.arange(4)
#a + np.array([1, 2]) # ValueError (모양이 맞지 않음)
```

```python
# Transposition
# triu() : 주어진 행렬의 상삼각행렬을 반환한다. (대각선 위의 요소만 1, 나머지는 0)
a = np.triu(np.ones((3, 3)), 1)
a
"""
풀이
1. np.ones((3, 3))은 3x3 행렬을 생성하고 모든 요소를 1로 초기화한다.
2. np.triu()는 주어진 행렬의 상삼각행렬을 반환한다. (대각선 위의 요소만 1, 나머지는 0)
3. 따라서 결과는 다음과 같다.
   [[0. 1. 1.]
    [0. 0. 1.]
    [0. 0. 0.]]
"""
```

```python
# triu()의 전치행렬 (view)
a.T
```
