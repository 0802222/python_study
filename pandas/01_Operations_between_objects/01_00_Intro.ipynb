# Pandas.DataFrame 클래스 기본

### DataFrame
: 행과 열에 이름이 있고, 열마다 서로 다른 자료형을 담을 수 있는 2차원 표 형태의 데이터 구조

- 딕셔너리의 key 를 열의 이름으로, list 를 열의 데이터로 만들 수 있다.
- 행 이름을 지정하지 않으면, 0, 1, 2, ... 처럼 기본으로 붙는다.

```python
import pandas as pd

# DataFrame 생성 (컬럼 지정)
df = pd.DataFrame({
    "이름": ["영희", "민수"],
    "나이": [33, 28]
})

print(df)
```

```python
# DataFrame 생성 (index 지정)
df = pd.DataFrame({
    "이름": ["영희", "민수"],
    "나이": [33, 28]
}, 
index=["학생1", "학생2"])

print(df)
```

```python
# 클래스 구조
class pandas.DataFrame(data=None, index=None, columns=None, dtype=None, copy=False)

    """
    - pd.DataFrame()으로 생성된 인스턴스는 크기변경이 가능한 2차원 배열이다.
      (Series는 1차원 배열이다.)
    - 데이터구조에는 레이블이 지정된 축(행, 열)이 있고, 클래스 메서드를 통해 레이블의 수정이 가능하다.  
    """
    
# Parameters
    """
    - data : 다양한 형태의 데이터 입력 가능 (ndarray, Series, DataFrame, dict, list, iterable)
    - index : index 또는 배열 형태의 객체 (행 레이블)
    - columns : index 또는 배열 형태의 객체 (열 레이블)
    - dtype : 데이터 타입을 강제하고자 할때 사용 (기본값 : None, None이면 자동으로 데이터 타입을 추론)
    - copy : 데이터 복사 여부 (기본값 : False 원본배열 수정, True이면 데이터 복사)
    """
```

```python
# copy() 예제
import numpy as np
np.random.seed(0) # seed 를 설정하면 동일한 난수 생성 가능
arr = np.random.randint(10, size=(2, 2)) # 2x2 배열 생성
print(arr)

df1 = pd.DataFrame(arr, copy = False) # copy = False, 원본 배열 수정
df2 = pd.DataFrame(arr, copy = True) # copy = True, 데이터 복사

arr[0, 0] = 99 
print("df1 : \n", df1) # 원본 배열 수정(99로 바뀜)
print("df2 : \n", df2)
```

```python
# 예시 01 - dictionary 로 만들기
# 데이터 자체가 2차원 배열일 경우, data만 설정해도 DataFrame 객체 생성 가능

data = {'A': [1, 2], 'B': [3, 4]}
df = pd.DataFrame(data = data)
print(df)
```

```python
# 예시 02 - index, columns 지정
data = np.array([[1,2], [3,4]])

df = pd.DataFrame(data = data, 
                  index = ['row1', 'row2'], 
                  columns = ['col1', 'col2'])
print(df)
```
