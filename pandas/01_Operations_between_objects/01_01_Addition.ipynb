## 01-01. 덧셈
### add()
: DataFrame에 다른 데이터프레임이나, Series, 스칼라 등 데이터를 더하는 메서드이다.
단순 df + 다른 df 의 계산과 차별화 되는것은, fill_value 인수를 통해 계산 불가한 값을 채워넣는다는 것이다.

`DataFrame.add(other, axis='columns', level=None, fill_value=None)`

### radd()
: radd 는 add의 경우에서 순서만 바꾼것이다.
즉 add 가 df+df2라면 radd는 df2 + df 이다.
`DataFrame.radd(other, axis='columns', level=None, fill_value=None)`

#### 기본 사용법
- other : 다른 데이터프레임, Series, 스칼라 등의 데이터
- axis : 더할 레이블 설정 (0 은 index(행), 1 은 columns(열))
    - Series의 경우 index와 일치시킬 축을 의미
- level : multiIndex 에서 계산할 Index의 레벨
- fill_value : NaN 값 등의 누락 요소를 계산 전에 이 값으로 대체    

```python
# 예시 01 - 3x3 DataFrame 생성
data = [[1, 10, 100], [2, 20, 200], [3, 30, 300]]

col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3']

df = pd.DataFrame(data = data, index = row, columns = col)

print(df)
```

```python
# 스칼라 값 더하기
# df.add(스칼라 값) : df + 스칼라 값 과 같은 결과
result = df.add(1)
print(result)

result2 = df + 1
print(result2) # result1 과 같은 결과
```

```python
# 다른 DataFrame 객체 더하기
# df에 더할 df2 객체 생성
data2 = [[3], [4], [5]]

df2 = pd.DataFrame(data=data2, index = ['row1', 'row2', 'row3'], columns = ['col1'])

print(df2)
```

```python
# df.add(df2)
print(df)
print(df2)
result = df.add(df2)

print(result) # df2에는 col2, col3가 없으므로 NaN으로 채워짐
```

```python
# fill_value 인수를 통해 NaN을 다른 값으로 채워서 연산 가능
result = df.add(df2, fill_value=0)
print(result)
```
