# 01-02. 뺄셈(sub, rsub)

## sub()
: DataFrame 에 다른 데이터프레임이나, Series, 스칼라 등 데이터를 빼는 메서드이다.
덧셈과 마찬가지로 fill_value 인수를 통해 계산 불가한 값을 채워넣는것이 다른 df 뺄셈과 차별점이다.

## rsub()
: sub()의 경우에서 순서만 바꾼 것 이다. sub가 df-df2라면, rsub는 df2-df 이다.

```python
import pandas as pd
```

```python
# 3 x 3 DataFrame 생성
data = [[1, 10, 100], [2, 20, 200], [3, 30, 300]]
col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3']

df = pd.DataFrame(data = data, index = row, columns = col)

print(df)
```

```python
# 스칼라 값 빼기
result = df.sub(1) # df - 1 과 같다.
print(result)
```

```python
# 다른 DataFrame 빼기

# df2 생성
data2 = [[3], [4], [5]]
df2 = pd.DataFrame(data = data2, index = row, columns = ['col1'])

print(df2)

# df - df2 (df2에 없는 col2, col3는 NaN으로 처리)
result = df.sub(df2)
print(result)
```

```python
result = df.sub(df2, fill_value = 0) # fill_value = 0 옵션을 주면 NaN이 아닌 0으로 처리돼서 계산할 수 있음
print(result)
```
