# 01-03. 곱셈(mul, rmul)

## mul()
: DataFrame 에 다른 데이터프레임이나, Series, 스칼라 등 데이터를 곱하는 메서드이다.
앞선 메서드와 마찬가지로 fill_value 인수를 통해 계산 불가한 값을 채워넣는것이 다른 df 연산과 차별점이다.

## rmul()
: mul()의 경우에서 순서만 바꾼 것 이다. mul이 df * df2라면, rmul은 df2 * df 이다.

```python
import pandas as pd
```

```python
# 3 x 3 DataFrame 생성
data = [[1, 10, 100], [2, 20, 200], [3, 30, 300]]
col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3']

df = pd.DataFrame(data = data, index = row, columns = col)

print(df)
```

```python
# 스칼라 값 곱하기
result = df.mul(2) # df * 2 과 같다.
print(result)
```

```python
# 다른 DataFrame 곱하기
data2 = [[3], [4], [5]]
df2 = pd.DataFrame(data = data2, index = row, columns = ['col1'])

print(df2)

# df * df2 (df2에 없는 col2, col3은 NaN으로 표시됨)
result = df.mul(df2)
print(result)
```

```python
# fill_value 옵션을 주면 NaN이 아닌 0으로 처리돼서 계산할 수 있음
result = df.mul(df2, fill_value = 0)
print(result)
```
