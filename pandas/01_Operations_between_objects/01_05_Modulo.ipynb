# 01-05. 나머지(mod, rmod)

## mod()
: DataFrame 에 다른 데이터프레임이나, Series, 스칼라 등 데이터를 나눴을 때 나머지를 반환하는 메서드이다.
앞선 메서드와 마찬가지로 fill_value 인수를 통해 계산 불가한 값을 채워넣는것이 다른 df 연산과 차별점이다.

## rmod()
: mod()의 경우에서 순서만 바꾼 것 이다. mod가 df % df2라면, rmod는 df2 % df 이다.

```python
import pandas as pd
```

```python
data = [[1,2,3],[4,5,6],[7,8,9]]
col = ['col1','col2','col3']
row = ['row1','row2','row3']
df = pd.DataFrame(data=data,index=row,columns=col)
print(df)
```

```python
# 스칼라 값 나누기
result = df.mod(7) # df % 7 과 같다.
print(result)
```

```python
# 다른 DataFrame 객체를 나누기
data2 = [[2], [3], [5]]
df2 =pd.DataFrame(data = data2, index = ['row1', 'row2', 'row3'], columns = ['col1'])

print(df2)
```

```python
# 메서드를 통해 나머지 구하기
result = df.mod(df2)
print(result)
```

```python
# fill_value 옵션을 주면 NaN이 아닌 0으로 처리돼서 계산할 수 있음
result = df.mod(df2, fill_value = 1)
print(result)
```
