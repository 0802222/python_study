# 01-06. 거듭제곱(pow, rpow)

## pow()
: DataFrame 에 다른 데이터프레임이나, Series, 스칼라 등 데이터를 거듭제곱 하는 메서드이다.
앞선 메서드와 마찬가지로 fill_value 인수를 통해 계산 불가한 값을 채워넣는것이 다른 df 연산과 차별점이다.

## rpow()
: pow()의 경우에서 순서만 바꾼 것 이다. pow가 df ** df2라면, rpow는 df2 ** df 이다.

```python
import pandas as pd
```

```python
# 3 * 3 DataFrame 객체 생성
data = [[1,2,3],[4,5,6],[7,8,9]]
col = ['col1','col2','col3']
row = ['row1','row2','row3']

df = pd.DataFrame(data=data,index=row,columns=col)

print(df)
```

```python
result = df.pow(3) # df ** 3 과 같다.
print(result)
```

```python
# DataFrame 제곱하기
data2 = [[0], [3], [5]]
df2 = pd.DataFrame(data = data2, index = row, columns = ['col1'])
print(df2)
```

```python
# 메서드를 통해 제곱하기
result = df.pow(df2)
print(result)
```

```python
# fill_value 옵션을 주면 NaN이 아닌 0으로 처리돼서 계산할 수 있음
result = df.pow(df2, fill_value = 1)
print(result)
```
