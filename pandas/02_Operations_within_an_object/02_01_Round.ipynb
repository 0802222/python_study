# 02-01. 반올림 (round)
: DataFrame 객체 내의 요소를 반올림 하는 메서드 이다.

`DataFrame.round(decimals=0, args, kwargs)`
- decimals : 소수 n 번째 자리 까지 반올림 한다. (음수면 10의 n승 자리까지 반올림)

```python
import pandas as pd
import numpy as np
```

```python
# 3 * 3 난수 DataFrame 객체 생성
col = ['col1','col2','col3']
row = ['row1','row2','row3']

data = np.random.rand(3,3)*100
df = pd.DataFrame(data=data, index=row, columns=col)

print(df)
```

```python
# decimals = 0 인 경우 반올림 (1의 자리까지 반올림)
print(df.round(0))
```

```python
# decimals > 0 인 경우 (양수, 소수 n번째 자리까지 반올림)
print(df.round(3))
```

```python
# decimals < 0 인 경우 (음수, 10의 n승 자리까지 반올림)
# ex) n이 -3이면 10의 -3승 자리까지 반올림한다는 의미로, 10의 -3승은 0.001이므로 소수점 셋째 자리까지 반올림한다는 의미이다.
print(df.round(-1)) # 소수점 첫째 자리까지 반올림
print(df.round(-2)) # 소수점 둘째 자리까지 반올림
print(df.round(-3)) # 소수점 셋째 자리까지 반올림 
```
