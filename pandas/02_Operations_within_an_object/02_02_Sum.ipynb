# 02-02. 합계(sum)

## sum()
: 객체의 행이나 열의 총 합계를 구하는 메서드이다.

### 사용법
```python
df.sum(axis=None, skipna=None, level=None, numeric_only=None, min_count=0)
```
- axis : { 0 : 행 / 1 : 열 } 더할 레이블을 선택한다.
- skipna : { True or False } NaN이 존재할 경우 무시할지의 여부이다. 기본값은 True이다.
- level : MultiIndex일 경우 레벨을 설정한다.
- numeric_only : 숫자 데이터만 사용할지의 여부이다.
- min_count : 계산에 필요한 숫자의 최소 개수이다.

````python
import numpy as np
import pandas as pd
````

````python
# NaN이 포함된 3 x 3 DataFrame 생성
col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3']
data = [[1, 2, 3], [4, 5, 6], [7, np.nan, 9]]

df = pd.DataFrame(data = data, index = row, columns = col)

print(df)
````

````python
# axis 를 설정하여 더하기 수행 (NaN 값은 무시하고 계산됨)
print(df.sum(axis = 0)) # 열 방향으로 더하기
print(df.sum(axis = 1)) # 행 방향으로 더하기
````

````python
# skipna (skipna=False 일 경우 NaN 값이 포함되어 있으면 NaN 반환)
print(df.sum(axis=0, skipna = False)) # 열 방향으로 더하기
print(df.sum(axis=1, skipna = False)) # 행 방향으로 더하기
````

````python
# min_count : 계산에 필요한 최소한의 유효한 값 개수 지정
# ex) min_count = 3 일때 NaN 을 포함하는 행의 경우 숫자가 2개이기 때문에 skipna = True 로 설정되어 있어도 NaN 반환
print(df.sum(axis=1))
print("==")
print(df.sum(axis=1, min_count = 3)) # row3 = 7 + NaN + 9 = NaN 반환 (숫자가 2개뿐이다.)
````
