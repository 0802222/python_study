# 02-03. 곱(prod, product)

## prod(), product()
: prod 메서드는 객체의 행이나 열의 곱을 구하는 메서드이다. product 메서드와 동일하다.

### 사용법
```python
df.prod(axis=None, skipna=None, level=None, numeric_only=None, min_count=0)
```
- axis : { 0 : 행 / 1 : 열 } 곱할 레이블을 선택한다.
- skipna : { True or False } NaN이 존재할 경우 무시할지의 여부이다. 기본값은 True이다.
- level : MultiIndex일 경우 레벨을 설정한다.
- numeric_only : 숫자 데이터만 사용할지의 여부이다.
- min_count : 계산에 필요한 숫자의 최소 개수이다.

````python
import numpy as np
import pandas as pd
````

````python
# NaN이 포함된 3 x 3 DataFrame 생성
col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3']
data = [[1, 2, 3], [4, 5, 6], [7, np.nan, 9]]

df = pd.DataFrame(data = data, index = row, columns = col)

print(df)
````

````python
# axis 설정하여 계산 (NaN 값은 무시하고 계산됨)
print(df.prod(axis = 0)) # 열 방향으로 곱하기
print(df.prod(axis = 1)) # 행 방향으로 곱하기
````

````python
# skipna 옵션을 False로 설정하면 NaN 값이 포함된 경우 NaN 반환
print(df.prod(axis = 0, skipna = False)) # NaN 값이 포함되어 있으므로 NaN 반환 ( col2 = 2 * 5 * NaN = NaN )
````

````python
# min_count 를 변경하여 계산
print(df.prod(axis = 1, min_count = 3)) # row3 = 7 * NaN * 9 = NaN 반환 (숫자가 2개뿐이다.)
````
