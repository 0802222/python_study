# 02-04. 절댓값(abs)

## abs()
: 각 요소의 절댓값을 반환하는 메서드이다.

### 사용법
```python
df.abs()
```
- 숫자의 경우 절댓값을 반환하며, 복소수의 경우 복소수의 크기가 반환된다.
    - 복소수가 a+bj인 경우 크기는 √(a² + b²) 이다.
- NaN의 경우는 NaN을 그대로 반환한다.

#### 복소수
: 복소수란 실수부와 허수부를 모두 갖는 수를 의미한다.
- 일반적으로 `a + bi` 형태로 나타낸다. 
- 여기서 a는 실수부, b는 허수부, i는 허수 단위로 정의된다. (ex. 3 + 4i에서 3은 실수부, 4는 허수부)

````python
import numpy as np
import pandas as pd
````

````python
# NaN과 복소수가 포함된 3 x 3 DataFrame 생성
# 여기서 복소수는 3 + 4j 이다.
col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3']
data = [[-1, 2, -3.5], [4, -5.5, 3+4j], [7, np.nan, 0]]

df = pd.DataFrame(data = data, index = row, columns = col)

# 실행했을 때 col3은 1개의 복소수로 인해 dtype이 complex128로 설정된다.
# (complex128은 복소수(complex number)를 저장할 수 있는 자료형이다.)
print(df)
print("==")
print(df.dtypes)
````

````python
# abs 메서드 실행 (일반적으로 절대값, 복소수의 경우 크기(magnitude) 계산)
print(df.abs())

# (1,2) 가 5.0인 이유
# |3 + 4j| = √(3² + 4²)
#          = √(9 + 16)
#          = √25
#          = 5
````
