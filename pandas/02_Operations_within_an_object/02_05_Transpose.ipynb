# 02-05. 전치(transpose, T)

## transpose(), T
: DataFrame 객체를 전치하는 메서드이다.
- (n, m) 크기의 DataFrame이라면 (0, 0)부터 (n, m)을 연결하는 대각선을 중심으로 뒤집는 것과 같다.
- T 속성은 transpose 메서드와 동일하다.

### 사용법
```python
df.transpose(*args, copy=False)
```
- copy : 사본을 반환할지 여부이다. 여러 dtype으로 이루어진 경우 자동으로 True가 된다.

````python
import pandas as pd
````

````python
# 4 x 3 DataFrame 생성
col = ['col1', 'col2', 'col3']
row = ['row1', 'row2', 'row3', 'row4']
data = [['A', 1, 2], ['B', 3, 4], ['C', 5, 6], ['D', 7, 8]]

df = pd.DataFrame(data = data, index = row, columns = col)

print(df)
````

````python
# transpose() 메서드 실행 :행과 열을 바꾸는 메서드
print(df.transpose())
````
