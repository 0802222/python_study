# 02-06. 순위(rank)

## rank()
: 축에 대해서 순위를 매기는 메서드이다. 
- 동일 순위일 경우 평균을 반환한다.

### 사용법
```python
df.rank(axis=0, method='average', numeric_only=None, na_option='keep', ascending=True, pct=False)
```
- axis : { 0 : index / 1 : columns } 순위를 매길 레이블이다.
- method : { 'average' / 'min' / 'max' / 'first' / 'dense' } 동순위일 때 처리 방법이다.
    - average : 평균
    - min : 낮은 순위
    - max : 높은 순위
    - first : 나타나는 순서대로
    - dense : min과 같지만 그룹 간 순위는 항상 1씩 증가
- numeric_only : { True / False } 숫자만 순위를 매길지 여부이다.
- na_option : { 'keep' / 'top' / 'bottom' } NaN 값의 처리 방법이다.
    - keep : NaN의 순위를 계산하지 않고 NaN으로 둔다.
    - top : NaN을 정렬 순서의 맨 앞에 둔다.
    - bottom : NaN을 정렬 순서의 맨 뒤에 둔다.
- ascending : { True / False } 오름차순으로 할지의 여부이다.
- pct : { True / False } 순위를 백분위수 형식으로 할지 여부이다.

````python
import pandas as pd
````

### `np.nan`과 `pd.NA`의 차이

둘 다 **결측값**을 나타내지만, 목적과 동작이 다릅니다.

| 구분 | `np.nan` | `pd.NA` |
|---|---|---|
| 소속 | NumPy | pandas |
| 의미 | Not a Number | pandas의 결측값 |
| 자료형 | `float` | `NAType` |
| 정수 열에서 사용 | 열 전체가 `float`로 바뀔 수 있음 | 정수형 유지 가능 |
| 비교 결과 | `False` | `pd.NA` |
| 논리 연산 | 일반적인 `True/False` 결과 | 결과가 불확실하면 `pd.NA` |

````python
# pd.NA가 포함된 9 x 1 DataFrame 생성
# 같은 수는 보기 편하도록 별 표시
data = [[5], [5], [pd.NA], [3], [-3.1], [5], [0.4], [6.7], [3]]
row = ['A★', 'B★', 'C', 'D☆', 'E', 'F★', 'G', 'H', 'I☆']

df = pd.DataFrame(data = data, index = row, columns = ['Value'])

print(df)
````

````python
# method에 따른 차이
df['average']=df['Value'].rank(method='average')
df['min']=df['Value'].rank(method='min')
df['max']=df['Value'].rank(method='max')
df['first']=df['Value'].rank(method='first')
df['dense']=df['Value'].rank(method='dense')
print(df)

print("""
average : D와 I의 경우 각각 3등 4등이기때문에 3.5 출력 \n
min : A, B, F의 경우 각각 5등 6등 7등으로 가장 낮은등수인 5 출력 \n
max : A, B, F의 경우 각각 5등 6등 7등으로 가장 높등수인 7 출력 \n
first : 동점일경우 위에서부터 매김 D와 I 각각 3등 4등 \n
dense : min처럼 동작하지만 등수가 순차적으로 증가 \n
""")
````

````python
# na_option에 따른 차이
df['keep']=df['Value'].rank(na_option='keep')
df['top']=df['Value'].rank(na_option='top')
df['bottom']=df['Value'].rank(na_option='bottom')
df['pct']=df['Value'].rank(pct=True)
print(df)

print("""
keep : NaN의 순위를 계산하지 않고 NaN으로 둡니다.
top : NaN을 정렬 순서의 맨 앞에 둡니다. 따라서 1등이 됩니다.
bottom : NaN을 정렬 순서의 맨 뒤에 둡니다. 따라서 9등이 됩니다.
pct : True일 경우 백분위수로 표시합니다.
      """)
````
