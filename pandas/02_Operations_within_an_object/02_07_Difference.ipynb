## diff()
: 한 객체 내에서 열과 열 / 행과 행의 차이를 출력하는 메서드이다.

### 사용법
```python
df.diff(periods=1, axis=0)
```
- axis : 비교할 축을 지정한다. axis=0인 경우 행끼리 비교하고, axis=1인 경우 열끼리 비교한다.
- periods : 비교할 간격을 지정한다. 기본은 +1로 바로 이전 값과 비교한다.

````python
import pandas as pd
````

````python
# 8 x 3 DataFrame 생성
a = [1, 2, 3, 4, 5, 6, 7, 8]
b = [1, 2, 4, 8, 16, 32, 64, 128]
c = [8, 7, 6, 5, 4, 3, 2, 1]

data = {"col1": a, "col2": b, "col3": c}
df = pd.DataFrame(data)

print(df)
````

````python
# axis 에 따라 계산 방향이 달라진다. (axis = 0 : 행 방향으로 계산, axis = 1 : 열 방향으로 계산)
print(df.diff(axis = 0)) # 행 방향으로 계산 (행 - 바로 전 행)
print("0 행은 비교할 값이 없으므로 NaN 반환")
````

````python
print(df.diff(axis = 1)) # 열 방향으로 계산 (열 - 바로 전 열)
print("0 열은 비교할 값이 없으므로 NaN 반환")
````

````python
# periods 옵션을 통해 몇 번째 전의 값과 비교할지 설정 가능
print(df.diff(periods = 3)) # 3번째 전의 값과 비교 (행 - 3번째 전 행)
print("0, 1, 2 행은 비교할 값이 없으므로 NaN 반환")
print("열3 - 열0의 값 .... 열7 - 열4의 값 출력")
````

````python
print(df.diff(periods = -2)) # -2번째 전의 값과 비교 (행 - -2번째 전 행)
print("마이너스 값은 뒤로 이동하여 비교하는 것과 같다. (행 - 2번째 후 행)")
````
