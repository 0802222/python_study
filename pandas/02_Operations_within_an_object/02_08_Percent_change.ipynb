# 02-08. 차이[백분률]

## pct_change()
: 한 객체 내에서 행과 행의 차이를 현재 값과의 백분율로 출력하는 메서드이다.
즉, (현재 행의 값 - 이전 행의 값) ÷ 이전 행의 값 을 의미한다.

### 사용법
```python
df.pct_change(periods=1, fill_method=None, freq=None)
```
- periods : 비교할 간격을 지정한다. 기본은 +1로 바로 이전 값과 비교한다.
- fill_method : 현재 pandas에서는 None만 사용할 수 있다. 결측치는 pct_change()를 호출하기 전에 ffill() 또는 bfill()로 처리한다.
- freq : 시계열 API에서 사용할 증분을 지정한다. (예: 'M' 또는 BDay())

````python
import numpy as np
import pandas as pd
````

````python
# 6 x 3 DataFrame 생성
a = [1, 1, 4, 4, 1, 1]
b = [1, 2, 4, 8, 16, 32]
c = [1, np.nan, np.nan, np.nan, 16, 64]

data = {"col1": a, "col2": b, "col3": c}
df = pd.DataFrame(data)

print(df)
````

````python
# 기본적으로 (현재행 - 이전행) / 이전행 을 의미한다.
print(df.pct_change()) # 행 방향으로 계산
print("첫행은 이전행이 없으므로 NaN출력")
````

````python
# periods 인수 사용
print(df.pct_change(periods = 2)) # 2번째 전의 값과 비교 (행 - 2번째 전 행)
````

````python
# 결측치는 먼저 bfill() 또는 ffill()로 대체한 뒤 pct_change()를 호출한다.
print(df.bfill().pct_change()) # 결측치가 있으면 바로 아래의 값으로 대체
print(df.ffill().pct_change()) # 결측치가 있으면 바로 위의 값으로 대체
````
