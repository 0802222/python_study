# 02-09. 누적 계산(expanding)

## expanding()
: 행이나 열의 값에 대해 누적으로 연산을 수행하는 메서드이다.
`df.expanding().sum()` 처럼 추가 메서드를 이용하여 연산을 수행한다.

### 사용법
```python
df.expanding(min_periods=1, method='single').추가메서드()
```
- min_periods : 연산을 수행할 요소의 최소 개수이다. 이보다 작으면 NaN을 출력한다.
- center : 미사용
- axis : 현재 pandas의 expanding()에서는 지원하지 않는다. 열 방향 누적은 `df.T.expanding().sum().T`처럼 전치해서 계산한다.
- method : { single / table } 연산을 한 줄씩 수행할지, 전체 테이블에 대해서 수행할지 여부이다.
    - 기본값은 'single'로 한 줄씩 연산을 수행한다.
    - 'table'을 사용하기 위해서는 numba 라이브러리가 설치되어 있어야 하며, 추가 연산 메서드에서 `engine='numba'`로 설정해주어야 한다.

````python
import numba # method='table' 옵션을 사용하기 위해 설치 필요
import pandas as pd
````

````python
# 4 x 2 DataFrame 생성
data = {'col1': [1, 2, 3, 4], 'col2': [3, 7, 5, 6]}
idx = ['row1', 'row2', 'row3', 'row4']

df = pd.DataFrame(data = data, index = idx)

print(df)
````

````python
# 기본적인 사용 : df.expanding().추가 연산메서드() -> 누적으로 연산 진행
print(df.expanding().sum()) # 누적합
````

````python
# min_period : 입력값 만큼 충족되지않으면, 연산을 하지않고 NaN 반환
print(df.expanding(min_periods = 4).sum())
````

````python
# 열 방향으로 누적하려면 DataFrame을 전치한 뒤 다시 전치한다.
print(df.T.expanding().sum().T)
````

````python
# method='table' 옵션을 사용하면, numba를 이용하여 연산을 테이블 단위로 롤링해 연산속도를 높일 수 있다.
# 추가 연산 메서드에 인수로 engine='numba'를 지정하면 된다.
print(df.expanding(method='table').sum(engine='numba'))
````
